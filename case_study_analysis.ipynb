```python
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
import statsmodels.api as sm
```

```python
nssal2000 = pd.read_stata("CaseStudyNSSAL/UKDA-5223-stata/NSSAL_2000.dta") 
nssal2000.head()
```

```python
nssal2010 = pd.read_stata("CaseStudyNSSAL/UKDA-7799-stata/NSSAL_2010.dta")
nssal2010.head()
```

```python
# Show all column names
print(nssal2000.columns.tolist())
```

```python
print(nssal2010.columns.tolist())
```

```python
# 1. Clean column names 
nssal2000.columns = nssal2000.columns.str.strip().str.lower()
nssal2010.columns = nssal2010.columns.str.strip().str.lower()
```

```python
# 2. Keep only important variables for NSSAL 2000
cols_2000 = [
    "sserial",
    "rsex",
    "dage",
    "agrp",
    "agrp2",
    "total_wt",
    "psu",
    "strata",
    "hetlife",
    "hettot",
    "het5yrs",
    "het5yrg",
    "het1yr",
    "het1yrg",
    "het3mnt",
    "het3mntg",
    "hetnewp",
    "hetnonew",
    "hetnew2",
    "hetnew3",
    "educ",
    "educ2",
    "educ3",
    "exams2",
    "marstat",
    "ms",
    "ms2",
    "ms3",
    "gor",
    "gor2",
    "london",
    "het",
    "hom",
    "scelig",
    "sexcomp"
]
cols_2000_existing = [c for c in cols_2000 if c in nssal2000.columns]
missing_2000 = [c for c in cols_2000 if c not in nssal2000.columns]

print("Missing in 2000:", missing_2000)

nssal2000_small = nssal2000[cols_2000_existing].copy()
```

```python
# 3. Keep only important variables for NSSAL 2010
cols_2010 = [
    "sin2",
    "rsex",
    "dage",
    "agrp",
    "agrp2",
    "total_wt",
    "psu_scrm",
    "strata",
    "hetlife",
    "hettot",
    "het5yrs",
    "het5yrg",
    "het1yr",
    "het1yrg",
    "het3mnt",
    "het3mntg",
    "hetnewp",
    "hetnonew",
    "hetnew2",
    "hetnew3",
    "educ",
    "educ2",
    "educ3",
    "exams2",
    "marstat",
    "ms",
    "ms2",
    "ms3",
    "gor",
    "gor2",
    "london",
    "het",
    "sam",
    "scelign2",
    "sexcomp2"
]

cols_2010_existing = [c for c in cols_2010 if c in nssal2010.columns]
missing_2010 = [c for c in cols_2010 if c not in nssal2010.columns]

print("Missing in 2010:", missing_2010)

nssal2010_small = nssal2010[cols_2010_existing].copy()
```

```python
# 4. Rename variables so both datasets match
nssal2000_small = nssal2000_small.rename(columns={
    "sserial": "id"
})

nssal2010_small = nssal2010_small.rename(columns={
    "sin2": "id",
    "psu_scrm": "psu",
    "sam": "hom",
    "scelign2": "scelig",
    "sexcomp2": "sexcomp"
})
```

```python
# 5. Add survey wave
nssal2000_small["survey_wave"] = "1999_2001"
nssal2010_small["survey_wave"] = "2010_2012"
```

```python
# 6. Combine both datasets
combined = pd.concat(
    [nssal2000_small, nssal2010_small],
    ignore_index=True,
    sort=False
)

combined.head()
```

```python
#7. Keep an even smaller final analysis dataset
analysis_cols = [
    "id",
    "survey_wave",
    "rsex",
    "dage",
    "agrp",
    "het1yr",
    "het1yrg",
    "het5yrs",
    "het5yrg",
    "hettot",
    "educ3",
    "ms2",
    "gor2",
    "total_wt",
    "psu",
    "strata"
]

analysis_cols_existing = [c for c in analysis_cols if c in combined.columns]

analysis_df = combined[analysis_cols_existing].copy()
analysis_df.head()
```

```python
# 8. Recode missing values
missing_map = {
    "hettot": [-1, 9],
    "het5yrs": [-1, 9995, 9996, 9997, 9999],
    "het5yrg": [-1, 9],
    "het1yrg": [-1, 9],
    "educ3": [9],
    "ms2": [9],
    "rsex": [-1,-9, -8, -2]
}

for col, values in missing_map.items():
    if col in analysis_df.columns:
        analysis_df[col] = analysis_df[col].replace(values, np.nan)
```

```python
analysis_df["het1yr_clean"] = analysis_df["het1yr"].replace({
    -1: np.nan,
    999: np.nan,
    "not answered": np.nan,
    "Not answered": np.nan,
    "not applicable": np.nan,
    "Not applicable": np.nan,

    # conservative lower-bound coding
    995: 1,
    996: 2,
    997: 3,
    "at least one": 1,
    "At least one": 1,
    "at least two": 2,
    "At least two": 2,
    "at least three": 3,
    "At least three": 3
})

analysis_df["het1yr_clean"] = pd.to_numeric(
    analysis_df["het1yr_clean"],
    errors="coerce"
)
```

```python
for col in ["educ3", "ms2", "gor2"]:
    analysis_df[col] = (
        analysis_df[col]
        .astype("string")
        .str.strip()
        .str.lower()
    )
```

```python
# convert pd.NA to np.nan first
analysis_df = analysis_df.replace({pd.NA: np.nan})

# clean text categories first
for col in ["educ3", "ms2", "gor2"]:
    analysis_df[col] = (
        analysis_df[col]
        .astype("object")
        .astype(str)
        .str.strip()
        .str.lower()
    )

# replace text missing labels
missing_labels = [
    "not answered",
    "not answered/don't know",
    "don't know",
    "nan",
    "<na>"
]

for col in ["educ3", "ms2", "gor2"]:
    analysis_df[col] = analysis_df[col].replace(missing_labels, np.nan)
```

```python
# 9. Restrict to comparable age range
analysis_df = analysis_df[(analysis_df["dage"] >= 16) & (analysis_df["dage"] <= 44)]
```

```python
# 10. Create readable labels
analysis_df["wave_label"] = analysis_df["survey_wave"].map({
    "1999_2001": "Natsal-2",
    "2010_2012": "Natsal-3"
})
```

```python
# 11. fix gender capitalization
analysis_df["sex_clean"] = analysis_df["rsex"].str.lower().str.strip()

print(analysis_df["sex_clean"].value_counts(dropna=False))
```

```python
# Create female dummy
analysis_df["female"] = analysis_df["sex_clean"].map({
    "male": 0,
    "female": 1
})

print(analysis_df["female"].value_counts(dropna=False))
```

```python
# Create year dummy
analysis_df["year2010"] = analysis_df["survey_wave"].map({
    "1999_2001": 0,
    "2010_2012": 1
})

print(analysis_df["year2010"].value_counts(dropna=False))
```

```python
# Check overdispersion
mean_y = analysis_df["het1yr_clean"].mean()
var_y = analysis_df["het1yr_clean"].var()

print("Mean:", mean_y)
print("Variance:", var_y)
print("Variance / Mean:", var_y / mean_y)
```

```python
model = smf.glm(
    formula="""
        het1yr_clean ~ C(year2010) * C(female)
        + dage + C(educ3) + C(ms2) + C(gor2)
    """,
    data=analysis_df,
    family=sm.families.NegativeBinomial(),
    freq_weights=analysis_df["total_wt"]
)

result = model.fit()
print(result.summary())
```
